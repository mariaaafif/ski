Goal is to get all resorts that hosted women's FIS giant slalom in 2025-2026 worldowide

```python
# get every event or resort ID from calendar for women's GS
import re, time
import requests
from bs4 import BeautifulSoup
import pandas as pd

headers = {"User-Agent": "Mozilla/5.0 (student project: ski-race-analysis)"}
calendar_url = "https://www.fis-ski.com/DB/general/calendar-results.html?eventselection=&place=&sectorcode=AL&seasoncode=2026&categorycode=FIS&disciplinecode=GS&gendercode=W&racedate=&racecodex=&nationcode=&seasonmonth=X-2026&saveselection=-1&seasonselection="
response = requests.get(calendar_url, headers=headers, timeout=30)
soup = BeautifulSoup(response.text, "html.parser")

event_ids = []
for link in soup.find_all("a", href=True):
    match = re.search(r"eventid=(\d+)", link["href"])
    if match and match.group(1) not in event_ids:
        event_ids.append(match.group(1))

print(len(event_ids))
print(event_ids[:5])
```

```python
# For one event, return the race IDs that are women's GS
def get_race_ids(eventid):
    url = f"https://www.fis-ski.com/DB/general/event-details.html?sectorcode=AL&eventid={eventid}&seasoncode=2026"
    response = requests.get(url, headers=headers, timeout=30)
    soup = BeautifulSoup(response.text, "html.parser")

    race_text = {}
    for link in soup.find_all("a", href=True):
        match = re.search(r"raceid=(\d+)", link["href"])
        if match:
            raceid = match.group(1)
            race_text[raceid] = race_text.get(raceid, "") + " " + link.get_text(" ", strip=True)

    return [raceid for raceid, text in race_text.items()
            if "Giant Slalom" in text and "W" in text.split()]

print(get_race_ids("57814"))
```

```python
# Visit every event and save all women's GS race IDs
all_race_ids = []
for i, eventid in enumerate(event_ids):
    all_race_ids += get_race_ids(eventid)
    time.sleep(2)
    print(f"{i+1}/{len(event_ids)} events done, {len(all_race_ids)} races so far")

pd.DataFrame({"raceid": all_race_ids}).to_csv("../data/raw/race_ids.csv", index=False)
```
